```python
gedit input.txt 
    (My name is Rutuja shinde.
     My name is Shreya.
     I am a Hadoop student.)

gedit mapper.py
     (#!/usr/bin/env python3
import sys

for line in sys.stdin:
    line = line.strip()
    words = line.lower().split()
    for word in words:
        print(f"{word}\t1"))

gedit reducer.py
     (#!/usr/bin/env python3
import sys

word_count = {}

for line in sys.stdin:
    word, count = line.strip().split()
    word_count[word] = word_count.get(word, 0) + int(count)

for word, count in word_count.items():
    print(f"{word}\t{count}")
)

cat input.txt | python3 mapper.py | sort | python3 reducer.py

## Run Using Hadoop Streaming

hdfs dfs -mkdir /input
hdfs dfs -put input.txt /input

hadoop jar /home/admin1/hadoop-3.3.1/share/hadoop/tools/lib/hadoop-streaming-3.3.1.jar \
-file mapper.py -mapper mapper.py \
-file reducer.py -reducer reducer.py \
-input /input/input.txt -output /output

## View Output in HDFS

hdfs dfs -ls /output
hdfs dfs -cat /output/part-00000

```

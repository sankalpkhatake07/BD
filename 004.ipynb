```python
Create a MapReduce Program to Determine Student Grades.
```

```python
gedit input.txt

(Rutuja 85
Shreya 72
Sonal 90
Shreyash 78
Mayuresh 88
Alisha 95
Harshada 65
Arya 59
Omkar 70
Prathmesh 62
)

gedit mapper.py
(#!/usr/bin/env python3
import sys

for line in sys.stdin:
    line = line.strip()
    if not line:
        continue
    try:
        name, marks = line.split()
        print(f"{name}\t{marks}")
    except ValueError:
        continue
)


gedit reducer.py
(#!/usr/bin/env python3
import sys

def get_grade(marks):
    marks = int(marks)
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    else:
        return "F"

for line in sys.stdin:
    line = line.strip()
    if not line:
        continue
    name, marks = line.split("\t")
    grade = get_grade(marks)
    print(f"{name}\tGrade: {grade}")
)

cat input.txt | python3 mapper.py | sort | python3 reducer.py

##Run on Hadoop Cluste
hdfs dfs -mkdir /grades_input
hdfs dfs -put input.txt /grades_input

hadoop jar /home/admin1/hadoop-3.3.1/share/hadoop/tools/lib/hadoop-streaming-3.3.1.jar \
-file mapper.py -mapper mapper.py \
-file reducer.py -reducer reducer.py \
-input /grades_input/input.txt -output /grades_output


##View Output
hdfs dfs -cat /grades_output/part-00000
```
